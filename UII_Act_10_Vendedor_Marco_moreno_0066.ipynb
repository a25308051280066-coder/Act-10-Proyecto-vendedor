```python
# Lista vacía
vendedores = []

# Diccionario vacío
losvendedores = {}

# ID inicial
id_actual = 1

while True:

    print("\n===== MENU VENDEDOR INTELIGENTE =====")
    print("1. Agregar vendedor")
    print("2. Ver vendedores")
    print("3. Buscar vendedor")
    print("4. Actualizar vendedor")
    print("5. Eliminar vendedor")
    print("6. Ver diccionario losvendedores")
    print("7. Salir")

    opcion = input("Selecciona una opción: ")

    if opcion == "1":

        nombre = input("Nombre del vendedor: ")
        producto = input("Producto que vende: ")
        ventas = input("Cantidad de ventas: ")
        zona = input("Zona de trabajo: ")

        vendedor = {
            "id": id_actual,
            "nombre": nombre,
            "producto": producto,
            "ventas": ventas,
            "zona": zona
        }

        vendedores.append(vendedor)

        losvendedores[id_actual] = vendedor

        print("Vendedor agregado correctamente")

        id_actual = id_actual + 1

    elif opcion == "2":

        if vendedores == []:

            print("No hay vendedores registrados")

        else:

            print("\n===== LISTA DE VENDEDORES =====")

            for v in vendedores:

                print("-------------------------")
                print("ID:", v["id"])
                print("Nombre:", v["nombre"])
                print("Producto:", v["producto"])
                print("Ventas:", v["ventas"])
                print("Zona:", v["zona"])

    elif opcion == "3":

        buscar = input("Ingresa el nombre del vendedor: ")

        encontrado = False

        for v in vendedores:

            if v["nombre"].lower() == buscar.lower():

                print("\nVENDEDOR ENCONTRADO")
                print("ID:", v["id"])
                print("Nombre:", v["nombre"])
                print("Producto:", v["producto"])
                print("Ventas:", v["ventas"])
                print("Zona:", v["zona"])

                encontrado = True

        if encontrado == False:

            print("El vendedor no existe")

    elif opcion == "4":

        actualizar = input("Nombre del vendedor a actualizar: ")

        actualizado = False

        for v in vendedores:

            if v["nombre"].lower() == actualizar.lower():

                print("Ingresa los nuevos datos")

                v["nombre"] = input("Nuevo nombre: ")
                v["producto"] = input("Nuevo producto: ")
                v["ventas"] = input("Nuevas ventas: ")
                v["zona"] = input("Nueva zona: ")

                losvendedores[v["id"]] = v

                print("Vendedor actualizado correctamente")

                actualizado = True

        if actualizado == False:

            print("No se encontró el vendedor")

    elif opcion == "5":

        eliminar = input("Nombre del vendedor a eliminar: ")

        eliminado = False

        for v in vendedores:

            if v["nombre"].lower() == eliminar.lower():

                vendedores.remove(v)

                del losvendedores[v["id"]]

                print("Vendedor eliminado correctamente")

                eliminado = True

                break

        if eliminado == False:

            print("No se encontró el vendedor")

    elif opcion == "6":

        if losvendedores == {}:

            print("El diccionario está vacío")

        else:

            print("\n===== DICCIONARIO LOSVENDEDORES =====")

            print(losvendedores)

    elif opcion == "7":

        print("Saliendo del sistema...")
        break

    else:

        print("Opción incorrecta")
```
